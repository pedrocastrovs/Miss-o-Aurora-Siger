# Sistema de Telemetria Espacial

Atividade acadêmica para simular a análise de telemetria de uma missão espacial.

**Observação:** as faixas de segurança deste projeto são fictícias e usadas apenas para fins acadêmicos.

## 1. Dados de telemetria

Serão analisados temperatura interna e externa, integridade estrutural, energia, pressão dos tanques e módulos críticos.

````python
telemetria = {
    "temperatura_interna": 24.5,
    "temperatura_externa": 42.0,
    "integridade_estrutural": 1,
    "energia": 92,
    "pressao_tanques": 305,
    "modulos": {
        "propulsao": "OK",
        "navegacao": "OK",
        "comunicacao": "OK",
        "suporte_vida": "OK"
    }
}

telemetria
````

## 2. Algoritmo de verificação

**Pseudocódigo**

```text
INICIO
  ler dados de telemetria
  criar lista de falhas

  se temperatura interna fora de 18 a 30: registrar falha
  se temperatura externa fora de -20 a 50: registrar falha
  se integridade estrutural != 1: registrar falha
  se energia < 80: registrar falha
  se pressao fora de 250 a 320: registrar falha
  se algum modulo critico != OK: registrar falha

  se nao houver falhas:
      exibir PRONTO PARA DECOLAR
  senao:
      exibir DECOLAGEM ABORTADA e listar falhas
FIM
```

````python
def verificar_telemetria(dados):
    falhas = []

    if not 18 <= dados["temperatura_interna"] <= 30:
        falhas.append("Temperatura interna fora da faixa segura")

    if not -20 <= dados["temperatura_externa"] <= 50:
        falhas.append("Temperatura externa fora da faixa segura")

    if dados["integridade_estrutural"] != 1:
        falhas.append("Falha na integridade estrutural")

    if dados["energia"] < 80:
        falhas.append("Energia abaixo do minimo de 80%")

    if not 250 <= dados["pressao_tanques"] <= 320:
        falhas.append("Pressao dos tanques fora da faixa segura")

    for modulo, status in dados["modulos"].items():
        if status != "OK":
            falhas.append(f"Falha no modulo: {modulo}")

    return falhas

falhas = verificar_telemetria(telemetria)

print("=== SISTEMA DE TELEMETRIA ESPACIAL ===")
print(f"Temperatura interna: {telemetria['temperatura_interna']} C")
print(f"Temperatura externa: {telemetria['temperatura_externa']} C")
print(f"Integridade estrutural: {telemetria['integridade_estrutural']}")
print(f"Energia: {telemetria['energia']}%")
print(f"Pressao dos tanques: {telemetria['pressao_tanques']} bar")
print("Modulos:")
for modulo, status in telemetria["modulos"].items():
    print(f"  - {modulo}: {status}")

if not falhas:
    print("\nRESULTADO: PRONTO PARA DECOLAR")
else:
    print("\nRESULTADO: DECOLAGEM ABORTADA")
    for falha in falhas:
        print("-", falha)
````

## 3. Teste de falha

O cenário abaixo demonstra que o sistema também aborta a decolagem quando um parâmetro obrigatório não é atendido.

````python
teste_falha = telemetria.copy()
teste_falha["energia"] = 60
falhas_teste = verificar_telemetria(teste_falha)

print("Energia do teste:", teste_falha["energia"], "%")
print("RESULTADO: DECOLAGEM ABORTADA" if falhas_teste else "RESULTADO: PRONTO PARA DECOLAR")
print("Falhas:", falhas_teste)
````

## 4. Análise energética

Valores simulados:

- Capacidade total: **500 kWh**
- Carga atual: **92%**
- Perdas energéticas: **8%**
- Consumo da decolagem: **120 kWh**
- Consumo médio após a decolagem: **35 kW**

Fórmulas:

1. Energia armazenada = capacidade x carga atual
2. Energia útil = energia armazenada - perdas
3. Energia restante = energia útil - consumo da decolagem
4. Autonomia = energia restante / consumo médio

````python
capacidade_total = 500
carga_atual = 92
perdas_percentual = 8
consumo_decolagem = 120
consumo_medio = 35

energia_armazenada = capacidade_total * (carga_atual / 100)
perdas = energia_armazenada * (perdas_percentual / 100)
energia_util = energia_armazenada - perdas
energia_restante = energia_util - consumo_decolagem
autonomia_horas = energia_restante / consumo_medio

print(f"Energia armazenada: {energia_armazenada:.2f} kWh")
print(f"Perdas: {perdas:.2f} kWh")
print(f"Energia util: {energia_util:.2f} kWh")
print(f"Energia apos a decolagem: {energia_restante:.2f} kWh")
print(f"Autonomia estimada: {autonomia_horas:.2f} horas")
````
